```python
import seaborn as sns

titanic = sns.load_dataset("titanic")

print("Titanic dataset loaded successfully!")
```

```python
# Check for missing values
titanic.isnull().sum()
```

```python
# Missing value percentage
missing_percentage = (titanic.isnull().sum() / len(titanic)) * 100
missing_percentage
```

```python
# Check number of duplicate rows
titanic.duplicated().sum()
```

```python
# Display duplicate rows
titanic[titanic.duplicated()]
```

```python
titanic.groupby("sex")["survived"].mean()
```

```python
titanic.groupby("class")["survived"].agg(["count", "sum", "mean"])
```

```python
titanic.groupby("class")["age"].mean()
```

```python
print("TITANIC EDA SUMMARY")
print("-------------------")

print("Total passengers:", len(titanic))
print("Total columns:", len(titanic.columns))
print("Missing values:", titanic.isnull().sum().sum())
print("Duplicate rows:", titanic.duplicated().sum())

print("\nSurvival rate by gender:")
print(titanic.groupby("sex")["survived"].mean())

print("\nSurvival rate by passenger class:")
print(titanic.groupby("class")["survived"].mean())
```
