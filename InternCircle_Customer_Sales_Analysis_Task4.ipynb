```python
import pandas as pd

# Load Superstore dataset with the correct encoding
df = pd.read_csv(
    "/content/ecommerce_data/Sample - Superstore.csv",
    encoding="latin1"
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
# Convert date columns to datetime

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print("Date columns converted successfully!")
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
# Create useful date-based features

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter

print("Date features created successfully!")

df[["Order Date", "Year", "Month", "Month Name", "Quarter"]].head()
```

```python
# Sales by Year

yearly_sales = df.groupby("Year")["Sales"].sum().sort_index()

print("Yearly Sales:")
print(yearly_sales)
```

```python
# Customer Segment Analysis

segment_summary = (
    df.groupby("Segment")
      .agg(
          Customers=("Customer ID", "nunique"),
          Orders=("Order ID", "nunique"),
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Quantity=("Quantity", "sum")
      )
      .sort_values("Sales", ascending=False)
)

print("Customer Segment Summary:")
segment_summary
```

```python
# Calculate Profit Margin

segment_summary["Profit Margin %"] = (
    segment_summary["Profit"] / segment_summary["Sales"] * 100
).round(2)

segment_summary
```

```python
# Top 10 Customers by Sales

top_customers = (
    df.groupby(["Customer ID", "Customer Name"])
      .agg(
          Orders=("Order ID", "nunique"),
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum")
      )
      .sort_values("Sales", ascending=False)
      .head(10)
)

print("Top 10 Customers:")
top_customers
```

```python
# Pivot Table: Sales by Customer Segment and Category

sales_pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Segment",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)

print("Sales Pivot Table:")
sales_pivot
```

```python
# Pivot Table: Sales by Region and Category

region_category_pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)

print("Region-Category Sales Pivot Table:")
region_category_pivot
```

```python
# Category-wise Sales

category_sales = (
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print("Category-wise Sales:")
print(category_sales)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
# Sales by Customer Segment

plt.figure(figsize=(8,5))

sns.barplot(
    x=segment_summary.index,
    y=segment_summary["Sales"]
)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

## Business Recommendations

Based on the analysis of the e-commerce sales dataset, the following business recommendations can be made:

1. Customer Segmentation:
   Focus marketing and retention strategies on the customer segment generating the highest sales.

2. Product Strategy:
   Prioritize high-performing product categories while identifying opportunities to improve the performance of lower-selling categories.

3. Revenue Trends:
   Use yearly and monthly sales trends to plan inventory, promotions, and marketing campaigns during high-demand periods.

4. Regional Strategy:
   Compare regional sales performance and allocate marketing resources toward regions with strong growth potential.

5. Customer Management:
   Identify high-value customers based on sales and profit contribution and develop suitable customer retention strategies.

6. Profitability:
   Sales volume should be evaluated together with profit margin so that business decisions focus on profitable growth rather than sales alone.

7. Data-Driven Decision Making:
   Regularly monitor sales, customer segments, product categories, and regional performance using dashboards and pivot tables.
